## Step 1 – Understand the Concept

-A recursive function is a function that calls itself to solve a problem.

-The function keeps calling itself with a smaller or simpler value until it reaches a stopping condition, called the base condition

-Recursive call → the function calls itself.

-Base condition → tells the function when to stop.

Real-time example: Countdown

Suppose we want to count down from 5:

```python
def countdown(number):
    if number==0:
        return

    print(number)
    countdown(number-1)

countdown(5)
```

### Explanation

The countdown() function is a recursive function because it calls itself inside the function.

The if number == 0 condition is the base condition. It stops the recursion when the number reaches 0.

The print(number) statement prints the current number.

The countdown(number - 1) statement calls the same function again with the number decreased by 1.

## Generators

A generator is a special type of function that produces values one at a time instead of producing all values at once.

Generators use the yield keyword.

Real-time example:

Imagine an application has 1 lakh customer records.

Instead of loading all 1 lakh records into memory:


```python
def numbers():
    yield 1
    yield 2
    yield 3

result=numbers()

print(next(result))
print(next(result))
```

### Explanation

The numbers() function is a generator function because it uses the yield keyword.

The yield keyword returns a value one at a time and pauses the function.

The result = numbers() statement creates a generator object.

The next() function gets the next value from the generator.

The first next(result) returns 1.

The second next(result) returns 2.

The function pauses after each yield and continues from where it stopped when next() is called again.

### Decorators

A decorator is a function that is used to add extra functionality to another function without changing the original function's code.

Real-time example:

Imagine you have a basic mobile phone.

Basic Phone
    ↓
Add Camera
    ↓
Add extra feature

```python
def decorator(function):
    def wrapper():
        print("Before")
        function()
        print("After")
    return wrapper

@decorator
def hello():
    print("Hello")

hello()
```

### Explanation

A decorator is used to add extra behavior to an existing function without changing its original code.

The decorator() function receives another function as an argument.

The wrapper() function adds extra code before and after the original function.

The @decorator syntax applies the decorator to the hello() function.

When hello() is called, the wrapper() function runs.

First, "Before" is printed, then the original hello() function runs, and finally "After" is printed.

### Closures

```python
def create_greeting(name):
    def greeting():
        print("Hello",name)
    return greeting

hello=create_greeting("Dhinesh")

hello()
```

A closure is created when an inner function remembers and uses a value from the outer function.

The create_greeting() function takes name as an argument.

The inner greeting() function uses the name variable from the outer function.

The create_greeting("Dhinesh") statement creates the greeting() function and remembers the value "Dhinesh".

The hello() statement calls the returned function.

### global

global is used inside a function when we want to change a variable that belongs to the global scope

```python
count=0

def increase():
    global count
    count=count+1

increase()

print(count)
```

### Explanation

The count variable is created outside the function, so it is a global variable.

The increase() function uses the global keyword to access and modify the global count variable.

Without the global keyword, Python would treat count inside the function as a local variable when assigning to it.

The count = count + 1 statement increases the value of count by 1.

After calling increase(), the value of count becomes 1.

### nonlocal

nonlocal is used inside a nested function when we want to change a variable that belongs to the outer function.


```python
def counter():
    count=0

    def increase():
        nonlocal count
        count=count+1
        print(count)

    increase()
    increase()

counter()
```

### Explanation

The count variable is created inside the counter() function, so it belongs to the outer function.

The increase() function is an inner function.

The nonlocal keyword allows the increase() function to modify the count variable from the outer counter() function.

Each time increase() is called, the value of count increases by 1.

The increase() function is called two times, so the values 1 and 2 are printed.

### map()

map() is a built-in Python function used to apply the same function to each item in a sequence.

Real-world example:

Think of three students' marks. If we want to add 5 bonus marks to every student's mark, we can use map() to apply the same calculation to all students' marks.

```python
marks=[50,60,70]

def add_bonus(mark):
    return mark+5

result=map(add_bonus,marks)

print(list(result))
```

### Explanation

The marks list contains three marks: 50, 60, and 70.

The add_bonus() function adds 5 to each mark.

The map() function applies the add_bonus() function to every value in the marks list.

The result stores the mapped values.

The list() function converts the map object into a list.

### filter()

filter() is a built-in function that selects items from an iterable based on a condition.

It checks each item one by one. If the condition is True, the item is included in the result. If it is False, the item is excluded.

Real-world example

Imagine a teacher has marks of several students and wants to find only the students who scored 50 or more. filter() can check each mark and keep only the marks that satisfy this condition.

```python
marks=[40,55,30,80,65]

def passed(mark):
    return mark>=50

result=filter(passed,marks)

print(list(result))
```

### Explanation

The marks list contains different student marks.

The passed() function checks whether a mark is greater than or equal to 50.

The filter() function applies the passed() function to each mark.

Only the marks for which the condition is True are included in the result.

The list() function converts the filter object into a list.

### id()

id() is a built-in function used to get the unique identity of an object

Real-world example:

Think of a company employee. Two employees may have the same name, but each employee has a unique employee ID. Similarly, Python objects have their own identity, which we can see using id().

```python
name="Dhinesh"

print(id(name))
```

### Explanation

The name variable stores the value "Dhinesh".

The id() function returns the unique identity number of the object stored in the variable.

Here, id(name) returns the identity of the string object "Dhinesh".

The exact number may be different each time you run the program.
