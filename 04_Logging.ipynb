### What is Logging?

#### What is Logging?

Logging is the process of recording important information about what is happening inside a Python application while it is running.

Logging can record:

    Normal application activities
    Important events
    Warnings
    Errors
    Information useful for troubleshooting

### Why Use Logging?

Logging is used to record and monitor what is happening inside a Python application while it is running.

It is useful because developers can use the recorded logs to understand the application's behavior and troubleshoot problems.

#### Main Reasons to Use Logging
##### To monitor application activity
Logging records important activities such as application startup, database connection, user login, and data processing.
##### To find and troubleshoot errors
When an error occurs, logs can provide information about what happened and help developers identify the problem.
##### To track application flow
Logs help developers understand the sequence of operations performed by the application.
##### To monitor applications in production
Applications running on servers may not have a developer watching the terminal. Logs provide a record that can be checked later.
##### To record warnings and unusual situations
Logging can record situations that are not errors but may require attention, such as slow database responses or low disk space.

### print() vs logging

#### 1. print()

print() is mainly used to display information directly to the console.

```python
print("Application started")
print("Database connected")
```

#### 2. logging

logging is specifically designed to record and manage information about an application's execution

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.info("Database connected")
```

### Log Levels
#### Step 1 – Understand the Concept

#### 1. DEBUG

DEBUG is used for detailed information that helps developers understand and debug the application.

#### 2. INFO

INFO is used to record normal and important events in the application.

#### 3. WARNING

WARNING is used when something unexpected or potentially problematic happens, but the application can continue running.

#### 4. ERROR

ERROR is used when an operation fails or an error occurs, but the application itself may still continue running.

#### 5. CRITICAL

CRITICAL is used for a very serious problem that may prevent the application or an important part of the system from continuing to operate.

#Step 2 – Demonstrate the Concept

Real-World Example: Car

Imagine a car that monitors its different systems.

    DEBUG: The car records detailed engine information, such as engine temperature.
    INFO: The car records normal activities, such as the car starting successfully.
    WARNING: The car records a situation that needs attention, such as low fuel.
    ERROR: The car records a failed operation, such as the navigation system not working.
    CRITICAL: The car records a serious problem, such as a brake system failure.

```python
import logging

logging.basicConfig(level=logging.DEBUG, force=True)

logging.debug("Engine temperature is 90°C")
logging.info("Car started successfully")
logging.warning("Fuel level is low")
logging.error("Navigation system failed")
logging.critical("Brake system failure")
```

### Explanation

This program uses Python's logging module to record different car events.

    import logging imports the logging module.
    logging.basicConfig() configures the logging system.
    level=logging.DEBUG allows all log levels from DEBUG to CRITICAL.
    logging.debug() records detailed information.
    logging.info() records normal activities.
    logging.warning() records situations that need attention.
    logging.error() records failed operations.
    logging.critical() records serious problems.

#### Log Formatters

A Log Formatter is used to define the format of a log message.

Without a formatter, the output may look like:

        INFO:root:Car started successfully

With a formatter, we can include useful information such as:

    Date and time
    Log level
    Logger name
    Log message

For example:
    2026-09-28 07:30:15 - INFO - Car started successfully

### Step 2 – Demonstrate the Concept
        
Real-world example: Car

Suppose a car application records this event:
    Car started successfully

A formatter can add more information:
    2026-09-28 07:30:15 - INFO - Car started successfully

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.info("Car started successfully")
logging.warning("Fuel level is low")
logging.error("Navigation system failed")
```

Explanation of the Program

This program uses Python's logging module to create formatted log messages.

    import logging imports the logging module.
    logging.basicConfig() configures the logging system.
    level=logging.DEBUG allows all log levels from DEBUG to CRITICAL.
    format defines how the log messages are displayed.
    %(asctime)s displays the date and time.
    %(levelname)s displays the log level.
    %(message)s displays the actual log message.
    force=True applies the logging configuration in Jupyter Notebook.
    logging.info() records a normal information message.
    logging.warning() records a warning message.
    logging.error() records an error message.

### File Handlers

### Step 1 – Understand the Concept

A File Handler is used to save log messages into a file instead of displaying them only on the screen.

Normally:
    Python Program → Log Message → Console

With a File Handler:
    Python Program → Log Message → File

### Step 2 – Demonstrate the Concept

Real-world example: Car

Suppose a car application records into a files:
    car.log
The file may contain:
    INFO - Car started successfully
    WARNING - Fuel level is low
    ERROR - Navigation system failed

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    filename="car.log",
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.info("Car started successfully")
logging.warning("Fuel level is low")
logging.error("Navigation system failed")

with open("car.log", "r") as file:
    print(file.read())
```

Explanation 

This program uses Python's logging module to create log messages and save them into a file named car.log.

    import logging imports the logging module.
    logging.basicConfig() configures the logging system.
    level=logging.DEBUG allows all log levels from DEBUG to CRITICAL.
    filename="car.log" saves the log messages in the car.log file.
    format defines how the log messages are stored, including the date and time, log level, and message.
    force=True applies the logging configuration in Jupyter Notebook.
    logging.info() records that the car started successfully.
    logging.warning() records a warning that the fuel level is low.
    logging.error() records that the navigation system failed.
    with open("car.log", "r") as file: opens the car.log file in read mode.
    file.read() reads all the saved log messages from the file.
    print() displays the saved log messages in the Jupyter Notebook.

### Console Handlers


#### Step 1 – Understand the Concept

A Console Handler is used to send log messages to the console or screen.

When a Python program creates a log message, the Console Handler displays that message in the terminal.

#### Step 2 – Demonstrate the Concept
    
Real-world example: Car

Suppose a car application wants to show its current activities on the screen.

    INFO - Car started successfully
    WARNING - Fuel level is low
    ERROR - Navigation system failed

These messages are displayed directly on the screen.

The Console Handler is responsible for sending the log messages to the console.

So, if the developer is running the car application and wants to see what is happening right now, a Console Handler is useful.

```python
import logging

logger = logging.getLogger("car")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)

logger.addHandler(console_handler)

logger.info("Car started successfully")
logger.warning("Fuel level is low")
logger.error("Navigation system failed")
```

#### Explanation

    import logging imports Python's logging module.
    getLogger("car") creates a logger named car.
    setLevel(logging.DEBUG) allows all log levels from DEBUG to CRITICAL.
    logging.StreamHandler() creates a Console Handler.
    logging.Formatter() defines the format of the log messages.
    setFormatter() connects the formatter to the Console Handler.
    addHandler() connects the Console Handler to the logger.
    logger.info() records an information message.
    logger.warning() records a warning message.
    logger.error() records an error message.

### Rotating Logs (Bonus)

### Step 1 – Understand the Concept

Rotating Logs means automatically creating a new log file when the current log file becomes too large.

### Step 2 – Demonstrate the Concept
    
Real-world example: Car

```python
import logging

logging.basicConfig(
    filename="car.log",
    level=logging.WARNING,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.warning("Car fuel level is low")
logging.warning("Car needs service")
logging.warning("Car engine temperature is high")

print("Logs saved successfully.")
```

```python
Explanation

    import logging → Imports Python's built-in logging module.
    logging.basicConfig() → Configures the logging system.
    filename="car.log" → Saves the log messages in the car.log file.
    level=logging.WARNING → Records WARNING, ERROR, and CRITICAL messages.
    format=... → Defines the format of the log messages.
    %(asctime)s → Shows the date and time.
    %(levelname)s → Shows the logging level.
    %(message)s → Shows the actual log message.
    force=True → Applies the logging configuration in Jupyter Notebook.
    logging.warning() → Creates warning messages and saves them in the log file.
    print() → Displays a confirmation message on the screen.
```

```python
with open("car.log", "r") as file:
    print(file.read())
```

### explanation

    open("car.log", "r") → Opens the log file in read mode.
    with ... as file → Safely handles the file.
    file.read() → Reads all the content from the file.
    print() → Displays the saved logs.
