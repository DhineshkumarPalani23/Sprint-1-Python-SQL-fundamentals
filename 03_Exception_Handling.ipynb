### try and except

### Step 1 – Understand the Concept

The try block contains the code that may cause an exception. If an exception occurs inside the try block, Python transfers control to the except block. If no exception occurs, the try block executes successfully and the except block is skipped.

### Step 2 – Demonstrate the Concept
Real-world example

We try to open a door. If the door opens, we continue normally. If the door does not open, we handle the problem.

### Step 3 – Implement the Concept

```python
try:
    x = 10 / 0
    print(x)
except:
    print("An exception occurred.")
```

### Explanation

try-except is used to handle errors in Python.

The try block contains code that may cause an error.

Here, 10 / 0 causes a ZeroDivisionError because division by zero is not allowed.

When the error occurs, Python skips the remaining statements in the try block.

The except block catches the error and prints "An exception occurred.".

### else

### Step 1 – Understand the Concept

else block is used with try-except and executes only when no exception occurs in the try block. If an exception is raised, the else block is skipped.
    
### Step 2 – Demonstrate the Concept

Real-world example

We try to open a door. If the door opens successfully, we enter the room. If the door does not open, we handle the problem.

### Step 3 – Implement the Concept

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Please enter a valid number.")

else:
    print("You entered:", number)
```

### Explanation for Jupyter

The try block tries to convert the input into an integer.

If the user enters an invalid value, ValueError occurs.

The except block handles the error and displays a message.

The else block runs only when there is no error in the try block.

### finally 

### Step 1 – Understand the Concept

The finally block always executes, whether an exception occurs or not

### Step 2 – Demonstrate the Concept

Real-world Example:

After using a room, we clean the room whether everything went well or a problem occurred.

### Step 3 – Implement the Concept

```python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)

except ValueError:
    print("Please enter a valid number.")

finally:
    print("Program completed.")
```

### Explanation

The try block tries to convert the input into an integer.

If the user enters an invalid value, ValueError occurs.

The except block handles the error.

The finally block runs whether an error occurs or not.

### raise

### Step 1 – Understand the Concept

The raise keyword is used to manually raise an exception when a specific condition is not satisfied.

If the condition is satisfied, the program continues normally.

We can use raise with a specific exception such as ValueError.

### Real-World Example

If an employee's age is below 18, the program can raise an error because the age is not valid for the condition

### Step 3 – Implement the Concept

```python
age=15

if age<18:
    raise ValueError("Age must be 18 or above")

print("Employee is eligible")
```

### Explanation

age=15 → stores the employee's age.

if age<18: → checks whether the age is below 18.

raise ValueError(...) → manually creates an error because the condition is invalid.

print() is not executed because the exception stops the program.

### assert

### Step 1 – Understand the Concept

assert is used in Python to check whether a condition is true.

-If the condition is True, the program continues.

-If the condition is False, Python raises an AssertionError.

### Step 2 – Demonstrate the Concept
    
Real-World Example

We can use assert to check that an employee's age is valid.

### Step 3 – Implement the Concept

```python
age=20

assert age>=18,"Employee age must be 18 or above"

print("Valid employee age")
```

### Explanation

age=12 → stores the age as 12.

assert age>=18 → checks whether the age is 18 or above.

12>=18 is False.

Therefore, Python raises an AssertionError with the given message.

The print() statement does not execute because the assertion failed.


### Custom Exception

### Step 1 – Understand the Concept (Markdown)

What is a Custom Exception?

A Custom Exception is an exception that we create ourselves for a specific situation in our program.

Python already has built-in exceptions like:

1.ValueError

2.TypeError

3.ZeroDivisionError

But sometimes we need our own meaningful error. In that case, we can create a custom exception class.

### Step 2 – Demonstrate the Concept (Markdown)

Real-world example

Suppose a company has a rule that an employee must be at least 18 years old.

If someone enters an age below 18, we can create our own exception called InvalidAgeError.

### Step 3 – Implement the Concept

```python
class InvalidAgeError(Exception):
    pass


age=16

if age<18:
    raise InvalidAgeError("Employee age must be 18 or above")

print("Employee is eligible")
```

### Explanation

class InvalidAgeError(Exception): → creates our own custom exception.
    
Exception → is the built-in parent class for exceptions.
    
age=16 → stores the employee's age as 16.

if age<18: → checks whether the age is below 18.
    
raise InvalidAgeError(...) → manually raises our custom exception.
    
The print() statement does not execute because the exception occurs first.

### TypeError

### Step 1 – Understand the Concept (Markdown)

TypeError is a built-in Python exception that occurs when a function or operation receives an inappropriate data type.

#### Step 2 – Demonstrate the Concept (Markdown)

Real-world example

Suppose a program tries to add an employee's salary with their name.

Salary is a number, but name is a string.

The problem is the data type, not just the value.


```python
salary=25000
name="Dhinesh"

result=salary+name

print(result)
```

### Explanation
 
salary=25000 → stores an integer value.
    
name="Dhinesh" → stores a string value.
    
salary+name → tries to add an integer and a string.
    
Python cannot perform this operation because the data types are incompatible.
    
Therefore, Python raises a TypeError.

### Step 3 – Handle the Exception

```python
try:
    salary=25000
    name="Dhinesh"

    result=salary+name

    print(result)

except TypeError:
    print("Cannot add salary and name because they have different data types")
```

### Explanation

try → Python tries to perform the addition.
    
Integer and string cannot be added together.
    
Python raises a TypeError.
    
except TypeError catches the error.
    
The program displays a user-friendly message instead of stopping.

### ZeroDivisionError

### Step 1 – Understand the Concept

ZeroDivisionError is a built-in Python exception that occurs when we try to divide a number by zero.

Division by zero is not allowed in Python.

### Step 2 – Demonstrate the Concept

### Real-world example
    
Suppose a company wants to calculate the average salary of employees.

If the number of employees is 0, the program would try to divide the total salary by zero.

Therefore, Python raises a ZeroDivisionError.
    

```python
total_salary=50000
number_of_employees=0

average_salary=total_salary/number_of_employees

print(average_salary)
```

### Explanation

total_salary=50000 → stores the total salary.
    
number_of_employees=0 → stores the number of employees as 0.
    
total_salary/number_of_employees → tries to divide 50000 by 0.
    
Division by zero is not allowed.
    
Therefore, Python raises a ZeroDivisionError.
    
The print() statement does not execute.

### Step 3 – Handle the Exception

```python
try:
    total_salary=50000
    number_of_employees=0

    average_salary=total_salary/number_of_employees

    print(average_salary)

except ZeroDivisionError:
    print("Cannot calculate average salary because the number of employees is zero")
```

### Explanation

try → Python tries to calculate the average salary.
    
The program tries to divide by zero.
    
Python raises a ZeroDivisionError.
    
except ZeroDivisionError catches the error.
    
The program displays a user-friendly message instead of stopping.

### ImportError

### Step 1 – Understand the Concept

ImportError is a built-in Python exception that occurs when you try to import a required item that is not available in the module.

### Step 2 – Demonstrate the Concept

Real-world example

Suppose an AI/ML Engineer is working on a machine learning project and wants to import a specific function from a Python library.

If the library is available but the requested function is not available in that library, Python cannot import the function.

Therefore, Python raises an ImportError.

```python
from math import addition
```

### Explanation

from math import addition → tries to import addition from the math module.
    
The math module exists in Python.
    
But addition is not available in the math module.
    
Python cannot import the requested function.
    
Therefore, Python raises an ImportError

### Step 3 – Handle the Exception

```python
try:
    from math import addition
    print("Function imported successfully")

except ImportError:
    print("The required function could not be imported")
```

### Explanation

try → Python tries to import addition.

addition is not available in the math module.
    
Python raises an ImportError.
    
except ImportError catches the error.
    
The program displays a user-friendly message instead of stopping.
