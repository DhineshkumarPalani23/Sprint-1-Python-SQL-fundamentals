### Writing Text Files

### Step 1 – Understand the Concept

Writing a text file means creating a new text file or adding information to an existing text file using Python.

Python uses the open() function to open a file.
        "w"

Rules for Writing Text Files

    Use open() to open the file.
    Use "w" mode to write data to a file.
    "w" mode creates a new file if it does not exist.
    "w" mode replaces the existing content if the file already exists.
    Use write() to write text into the file.
    Use \n to write text on a new line.
    Use with to open and safely close the file.
    Use "a" mode when you want to add new content without removing existing content.

        "r" → Read
        "w" → Write / Replace
        "a" → Append / Add
### Step 2 – Demonstrate the Concept

Real-world example: Car

Suppose we want to save car information into a text file.

```python
with open("notes.txt" , "w") as car_info:
    car_info.write("Car Model: BMW\n")
    car_info.write("Fuel: Petrol\n")
    car_info.write("Status: Running\n")
print("Data written successfully.")
```

```python
Explanation:

    open("car.txt", "w") → Opens car.txt in write mode.
    "w" → Means write mode.
    with ... as file → Opens the file safely and automatically closes it after writing.
    file.write() → Writes text into the file.
    \n → Moves the next text to a new line.
    print() → Displays a confirmation message.
```

### Reading Text Files

#### Step 1 – Understand the Concept

Reading a text file means opening an existing text file and retrieving the information stored inside it.

Python provides the open() function to open a file and read its contents.

Rules for Reading Text Files

    Use the open() function to open a text file.
    Use "r" mode to read the file.
    "r" means read mode.
    The file should already exist when using "r" mode.
    Use read() to read the file content.
    Use with to safely open and automatically close the file.
    Store the file content in a variable if you need to use it later.
    Use print() to display the file content.

### Step 2 – Demonstrate the Concept
    
Real-world example

Suppose an AI/ML Engineer has a text file containing car information:

```python
with open("notes.txt" , "r") as file:
    content=file.read()
print(content)
```

#### Explanation
    open("notes.txt", "r") → Opens the notes.txt file in read mode.
    "r" → Means read mode. It is used to read the existing content of a file.
    with ... as file → Opens the file safely and automatically closes it after reading.
    file.read() → Reads the entire content of the notes.txt file.
    content = file.read() → Stores the content read from the file in the content variable.
    print(content) → Displays the content of the file on the screen.

### CSV Files

### Step 1 – Understand the Concept

CSV stands for Comma-Separated Values.

A CSV file is a simple file format used to store data in rows and columns.

Each value is separated by a comma ,.

Simple Rules for CSV Files

    Use the csv module to work with CSV files.
    Use csv.writer() to write data.
    Use csv.reader() to read data.
    Use writerow() to write one row.
    CSV data is stored in rows and columns.
    Values are usually separated by commas.
    Use "w" mode to create or overwrite a CSV file.
    Use "r" mode to read a CSV file.
    Use newline="" when opening a CSV file for writing.
    Use with to safely open and close the file.

### Step 2 – Demonstrate the Concept
    
Real-world example: Employee Data

Suppose a company wants to store employee information in a CSV file.

```python
import csv

with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Designation", "City"])
    writer.writerow(["Dhinesh", "AI/ML Engineer", "Salem"])

with open("employees.csv", "r") as file:
    content = csv.reader(file)

    for i in content:
        print(i)
```

Explanation

    import csv → Imports Python's built-in CSV module.
    open("employees.csv", "w", newline="") → Opens or creates the CSV file in write mode.
    csv.writer(file) → Creates a CSV writer object.
    writer.writerow() → Writes one row of data into the CSV file.
    ["Name", "Designation", "City"] → Creates the header row.
    ["Dhinesh", "AI/ML Engineer", "Salem"] → Writes employee information.
    open("employees.csv", "r") → Opens the CSV file in read mode.
    csv.reader(file) → Reads the CSV file row by row.
    for row in reader: → Goes through each row.
    print(row) → Displays each row.

### JSON Files

### Step 1 – Understand the Concept

JSON stands for JavaScript Object Notation.

JSON is a common format used to store and exchange data. It stores data using key-value pairs, similar to Python dictionaries.

Rules for JSON Files

    Use the json module to work with JSON files.
    Use json.dump() to write Python data to a JSON file.
    Use json.load() to read JSON data from a file.
    Use "w" mode to write/create a JSON file.
    Use "r" mode to read a JSON file.
    JSON data commonly uses key-value pairs.
    JSON keys are case-sensitive.
    Use with to safely open and close the file.

### Step 2 – Demonstrate the Concept
    
Real-world example: Employee Data

Suppose a company wants to store employee information.

```python
import json

data=[{
    "Name":"Dhineshkumar",
    "Designation":"AI/ML Engineer",
    "City":"Salem"
},
     {
         "Name":"Varun",
         "Designation":"Python Develper",
         "city":"chennai"
     }]

with open("employee.json" , "w") as file:
    json.dump(data,file)

with open("employee.json", "r") as file:
    content=json.load(file)
print(content)
```

Explanation

    import json → Imports Python's built-in JSON module.
    data → Stores employee information as a Python list of dictionaries.
    open("employee.json", "w") → Opens or creates the JSON file in write mode.
    json.dump(data, file) → Converts the Python data into JSON format and writes it to the file.
    open("employee.json", "r") → Opens the JSON file in read mode.
    json.load(file) → Reads the JSON data and converts it back into Python data.
    content → Stores the data read from the JSON file.
    print(content) → Displays the data on the screen.

### Pickle Files

### Step 1 – Understand the Concept

Pickle is a Python module used to save Python objects into a file and later load them back into Python.

It can store objects such as:
    
    Lists
    Dictionaries
    Tuples
    Classes and other Python objects

Pickle converts a Python object into a byte format for storage.

### Step 2 – Demonstrate the Concept

Real-world example:

Suppose an AI/ML Engineer has trained a model and wants to save the model for later use.

Instead of creating the model again every time, the model can be saved using Pickle and loaded when needed.

```python
import pickle

data={
    "Name":"Dhineshkumar",
    "Designation":"Python Develper",
    "City":"Salem"
}

with open("employee.pkl" , "wb") as file:
    pickle.dump(data,file)

with open("employee.pkl" , "rb") as file:
    content=pickle.load(file)

print(content)
```

Explanation

    import pickle → imports the Pickle module.
    data → stores employee information in a dictionary.
    "wb" → opens the file in write binary mode.
    pickle.dump(data, file) → saves the dictionary into employee.pkl.
    "rb" → opens the file in read binary mode.
    pickle.load(file) → loads the saved data back into Python.
    content → stores the loaded data.
    print(content) → displays the data.

### Context Manager (with)

Step 1 – Understand the Concept

A Context Manager is used to manage resources safely, such as files.

The with statement automatically opens the file and closes it after the work is completed, even if an error occurs.

Main benefit: We do not need to manually call file.close().

### Step 2 – Demonstrate the Concept

Real-world example:

Imagine you enter a meeting room:

    Enter Room → Use Room → Leave Room

You don't need to remember to manually lock everything each time. The process handles it safely.
    
    Open File → Read/Write → Automatically Close File

```python
with open("notes.txt", "r") as file:
    content = file.read()

print(content)
```

Explanation

    with → safely manages the file.
    open("notes.txt", "r") → opens notes.txt in read mode.
    "r" → means read.
    as file → gives the opened file a variable name, file.
    file.read() → reads the entire content of the file.
    content → stores the data read from the file.
    After the with block ends, the file is automatically closed.
    print(content) → displays the stored content.

### pathlib

Step 1 – Understand the Concept

pathlib is a built-in Python module used to work with files and directories easily.

It provides the Path class to create, check, and manage file and folder paths.

Main uses:

    Create file paths
    Check whether a file or folder exists
    Get file names
    Work with directories

Path()	Create a path
    
    .exists()	Check whether a path exists
    .is_file()	Check whether it is a file
    .is_dir()	Check whether it is a directory

### Step 2 – Demonstrate the Concept

Real-world example:

Suppose an AI/ML Engineer wants to check whether an employee.txt file exists before reading it.

Using pathlib, we can create a path and check it:

```python
from pathlib import Path

file_path = Path("employee.txt")

if file_path.exists():
    print("File exists")
else:
    print("File does not exist")
```

Explanation

    from pathlib import Path → imports the Path class.
    Path("employee.txt") → creates a path for the file.
    file_path → stores the file path.
    file_path.exists() → checks whether the file exists.
    if → checks the result.
    print() → displays the appropriate message.

### File Validation

### Step 1 – Understand the Concept

File Validation means checking whether a file is valid and suitable to use before performing an operation.

We can check:

Whether the file exists
Whether the path is actually a file
Whether the file has the correct extension
Whether the file is empty or not

### Step 2 – Demonstrate the Concept

Real-world example:

Before an AI/ML Engineer reads an employee.csv file, they can check:

```python
from pathlib import Path

file_path = Path("employees.csv")

if not file_path.exists():
    print("File does not exist.")
elif not file_path.is_file():
    print("The path is not a file.")
elif file_path.suffix != ".csv":
    print("Invalid file type.")
elif file_path.stat().st_size == 0:
    print("File is empty.")
else:
    print("File is valid.")
```

Explanation

    Path("employees.csv") → creates the file path.
    .exists() → checks whether the file exists.
    .is_file() → checks whether the path is a file.
    .suffix → gets the file extension.
    .stat().st_size → gets the file size.
    if / elif / else → performs the validation checks.
    If all checks are successful, it prints File is valid.

### Directory Operations

### Step 1 – Understand the Concept

Directory Operations means creating, checking, listing, renaming, and deleting folders (directories) using Python.

Python provides modules such as os and pathlib to work with directories.

### Step 2 – Demonstrate the Concept

Real-world example:

Suppose an AI/ML Engineer wants to create a folder called project_data to store project files.

```python
from pathlib import Path

folder = Path("project_data")

if not folder.exists():
    folder.mkdir()
    print("Directory created successfully.")
else:
    print("Directory already exists.")

print("Directory contents:")

for item in folder.iterdir():
    print(item.name)
```

Explanation
    
    Path("project_data") → creates a path for the directory.
    .exists() → checks whether the directory exists.
    .mkdir() → creates the directory.
    .iterdir() → gets the items inside the directory.
    item.name → gets the name of each file or folder.
    for loop → displays the directory contents
